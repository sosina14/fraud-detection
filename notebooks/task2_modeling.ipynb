```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold

# --- Fraud_Data pipeline ---
X_train_fraud = np.load('../data/processed/X_train.npy')
X_test_fraud  = np.load('../data/processed/X_test.npy')
y_train_fraud = np.load('../data/processed/y_train.npy')
y_test_fraud  = np.load('../data/processed/y_test.npy')

# --- creditcard pipeline (process separately) ---
cc_df = pd.read_csv('../data/raw/creditcard.csv')
cc_df.drop_duplicates(inplace=True)
cc_df.dropna(inplace=True)

X_cc = cc_df.drop(columns=['Class'])
y_cc = cc_df['Class']

X_train_cc, X_test_cc, y_train_cc, y_test_cc = train_test_split(
    X_cc, y_cc, test_size=0.2, stratify=y_cc, random_state=42
)

# Scale creditcard (already PCA'd, but still scale Amount and Time)
from sklearn.preprocessing import StandardScaler
scaler_cc = StandardScaler()
X_train_cc = scaler_cc.fit_transform(X_train_cc)
X_test_cc  = scaler_cc.transform(X_test_cc)

# SMOTE on creditcard train set only
from imblearn.over_sampling import SMOTE
sm = SMOTE(random_state=42)
X_train_cc_res, y_train_cc_res = sm.fit_resample(X_train_cc, y_train_cc)

print("Fraud_Data train shape:", X_train_fraud.shape)
print("creditcard train shape:", X_train_cc_res.shape)
```
